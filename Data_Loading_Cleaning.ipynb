```
import glob
import os

csv_files = glob.glob(
    "/content/arxiv_data/**/*.csv",
    recursive=True
)

matching_files = [
    file for file in csv_files
    if os.path.basename(file).lower()
    == "arxiv_scientific_dataset.csv"
]

if len(matching_files) == 0:
    raise FileNotFoundError(
        "The arXiv scientific dataset CSV was not found."
    )

csv_file = matching_files[0]

print("CSV file found successfully!")
print(os.path.basename(csv_file))
```

```
import pandas as pd

df = pd.read_csv(
    csv_file,
    low_memory=False
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```
# Check and remove complete duplicate rows

rows_before = len(df)
duplicate_rows = df.duplicated().sum()

print("Rows before duplicate removal:", rows_before)
print("Complete duplicate rows found:", duplicate_rows)

df = df.drop_duplicates().reset_index(drop=True)

rows_after = len(df)
rows_removed = rows_before - rows_after

print("Rows removed:", rows_removed)
print("Rows after duplicate removal:", rows_after)
print("Final shape after duplicate removal:", df.shape)

# Confirm that the calculated values are consistent
assert rows_removed == duplicate_rows, (
    "The number removed does not match the duplicate count."
)
```

```

# Remove duplicate rows if any
df = df.drop_duplicates()

print("Shape after removing duplicates:")
print(df.shape)
```

```
# Clean text columns without converting missing values to strings
text_columns = [
    "id",
    "title",
    "category",
    "category_code",
    "authors",
    "first_author",
    "summary"
]

for col in text_columns:
    # Convert blank spaces into real missing values
    df[col] = df[col].replace(r"^\s*$", np.nan, regex=True)

    # Remove spaces while preserving missing values
    df[col] = df[col].str.strip()

print("Text columns cleaned successfully.")
print(df.isnull().sum())
```

```
# Convert date columns to datetime
df["published_date"] = pd.to_datetime(
    df["published_date"],
    format="mixed",
    errors="coerce"
)

df["updated_date"] = pd.to_datetime(
    df["updated_date"],
    format="mixed",
    errors="coerce"
)

print(df[["published_date", "updated_date"]].dtypes)

print("\nMissing dates after conversion:")
print(df[["published_date", "updated_date"]].isnull().sum())
```

```
# Remove rows missing essential information
important_columns = [
    "id",
    "title",
    "category",
    "summary"
]

rows_before = len(df)

df = df.dropna(subset=important_columns).copy()

rows_removed = rows_before - len(df)

# Fill optional author information
df["authors"] = df["authors"].fillna("Unknown")
df["first_author"] = df["first_author"].fillna("Unknown")

# Recalculate summary word count
df["summary_word_count"] = (
    df["summary"]
    .str.split()
    .str.len()
)

df = df.reset_index(drop=True)

print("Rows removed:", rows_removed)
print("Final dataset shape:", df.shape)

print("\nRemaining missing values:")
print(df.isnull().sum())
```

```
# Save the newest cleaned dataset using a unique name
df.to_csv(
    "cleaned_arxiv_datasets.csv",
    index=False
)

# Reload it immediately to confirm the saved file
check_df = pd.read_csv(
    "cleaned_arxiv_datasets.csv",
    low_memory=False
)

print("Dataset before saving:", df.shape)
print("Saved file after reloading:", check_df.shape)
```
